## Цел на упражнението

Моделите от предишните части получаваха готови числа: размерите на
цветето или яркостта на пикселите. Много често данните не са така подходящи
за директно използване. Цената на акция е редица от стойности във времето, адресът е текст,
датата е низ като "2012-09-05 17:00", а SMS съобщението е изречение с
произволна дължина. Преди да се обучи какъвто и да е модел, данните трябва
да се разберат, почистят и превърнат в числа. Тази стъпка често определя
резултата повече от избора на модел.

Според формата си данните са три вида:

| | Структурирани | Полуструктурирани | Неструктурирани |
|---|---|---|---|
| Форма | таблица с фиксирани колони | записи с полета, но без фиксирана таблица | свободен текст, изображение, звук |
| Примери | CSV файл | обява във формат JSON | SMS съобщения |
| Подготовка | почистване, кодиране, мащабиране | полетата се превръщат в колони | извличане на признаци или вграждане |

Видовете данни в упражнението и как стават числа:

| Вид данни | Сурова форма | Числово представяне | За какво да се внимава |
|---|---|---|---|
| Времеви ред | стойности по дни | прозорец от 20 поредни дни | не трябва да се разбъркват |
| Адрес | низ | разстояние до центъра на града | измерване на различни неща |
| JSON | записи с вложени полета и списъци | колони от `pd.json_normalize` | липсващи полета и списъци с различна дължина |
| Дата и час | низ | час, ден, месец или синус и косинус | часът е цикличен: 23 е близо до 0 |
| Текст | низ с различна дължина | брой на всяка дума (Bag-of-Words) или TF-IDF | редът на думите се губи |

## Част 1. Времеви данни

**Времеви ред** е поредица от измервания на една и съща величина, направени
през равни интервали от време: цената на акция всеки борсов ден,
температурата всеки час, потреблението на ток всеки ден, пулсът всяка
секунда. Всяко измерване има свой момент във времето и **редът на
измерванията е част от данните**. Ако измерванията се разбъркат, се губи 
информацията, която е само в реда им: кое след кое се е случило 
(причинната структура) и кои ритми се повтарят (сезонната структура).

В таблицата с ирисите всяко цвете е отделен пример и редът на редовете няма
значение. Във времевия ред съседните стойности си приличат: днешното
потребление на ток е близо до вчерашното и до това от същия ден миналата
седмица. Тази зависимост помага за прогнозата, но изисква внимание при
подготовката на данните и при проверката на модела.

Преди модела се проверява следното:

| Свойство | Въпрос | Пример |
|---|---|---|
| честота | През колко време е всяко измерване? | всеки ден, всеки час |
| липсващи моменти | Има ли моменти без стойност и защо? | борсата не работи в събота |
| тренд | Има ли трайна посока нагоре или надолу? | цената на Apple расте с години |
| сезонност | Има ли ритъм, който се повтаря? | по-малко ток в неделя и през лятото |
| необичайни дни | Има ли скокове с външна причина? | разделяне на акциите, празници |

Най-честата задача е **прогнозата**: от миналите стойности да се предскаже
следващата. Тук са дадени два примера. Първо цените на акциите на Apple показват
какво се проверява и почиства в един ред. След това потреблението на ток във Франция се
превръща в таблица от **прозорци**: 20 поредни дни са признаците, а
21-вият ден е етикетът. Върху тази таблица се обучава линейна регресия.

### 1. Зареждане и оглед

`parse_dates` превръща низовете в колоната `date` в дати, а `index_col`
прави датата индекс на редовете. Така редовете могат да се избират по дата,
например `prices.loc["2005-02"]`.

````python
prices = pd.read_csv(DATA_DIR / "apple_stock.csv", parse_dates=["date"], index_col="date")
print("Форма:", prices.shape)
print("От", prices.index.min().date(), "до", prices.index.max().date())
display(prices.head())

fig, ax = plt.subplots(figsize=(10, 3.5), layout="constrained")
ax.plot(prices.index, prices["close"], color=BLUE, linewidth=1)
ax.set(ylabel="Цена при затваряне ($)", title="Apple, 2003-2007")
plt.show()
````

`open` е цената при отваряне на борсата, а `close` при затваряне. Графиката
показва **тренд**: цената расте почти през целия период. Има и странно
място: в началото на 2005 г. цената пада наполовина за един ден. Преди да
се търсят закономерности, такива места трябва да се разберат.

Първо честотата и липсващите моменти. `pd.infer_freq` познава честотата на
индекса, ако разстоянието между редовете е постоянно. `asfreq("D")` създава
ред за всеки календарен ден.

````python
print("Честота за целия период:", pd.infer_freq(prices.index))
print("Честота без празници:", pd.infer_freq(prices.loc["2003-01-06":"2003-01-17"].index))
print("Честота без почивни дни:", pd.infer_freq(prices.loc["2003-01-06":"2003-01-10"].index))
calendar = prices["close"].asfreq("D")
print("Календарни дни:", len(calendar), "| дни без цена:", int(calendar.isna().sum()))
print(calendar.loc["2003-01-01":"2003-01-12"])
````

`None` означава, че честотата не е постоянна. Борсата не работи в събота,
неделя и празници, затова за тези дни няма цена (`NaN`). Липсващата
стойност не е нула: цената не е паднала до 0, просто не е имало търговия.
Попълване с нули би създало несъществуващи сривове. Затова се работи само с
борсовите дни, по около 252 на година, а "вчера" означава предишния борсов
ден.

### 2. Почистване: разделяне на акциите

Къде е голямата промяна? `pct_change()` дава промяната спрямо предишния ред
като дял: -0.5 означава спад с 50%.

````python
daily_change = prices["close"].pct_change()
drop_day = daily_change.idxmin()
print("Най-голям дневен спад:", drop_day.date(), round(daily_change.loc[drop_day], 3))
position = prices.index.get_loc(drop_day) # позицията на този ден в таблицата
print(position)
display(prices.iloc[position - 3:position + 4]) # по 3 борсови дни преди и след спада
````

На 28 февруари 2005 г. Apple **разделя акциите си 2:1** (stock split):
всеки акционер получава две акции вместо една, а всяка струва наполовина.
Компанията не е загубила половината си стойност. Ако данните останат така,
моделът ще научи срив, който не се е случил. Затова цените преди
разделянето се делят на 2 и всички цени стават сравними.

Такава промяна не личи от самите числа. Нужно е знание за данните: какво е
измерено и какво се е случило. Корекцията се прави в копие, за да не се
повтори при ново изпълнение на клетката.

````python
SPLIT_DAY = pd.Timestamp("2005-02-28")
adjusted = prices.copy()
adjusted.loc[adjusted.index < SPLIT_DAY, ["open", "close"]] /= 2
print("Най-голяма дневна промяна след корекцията:", round(adjusted["close"].pct_change().abs().max(), 3))

fig, ax = plt.subplots(figsize=(10, 3.5), layout="constrained")
ax.plot(adjusted.index, adjusted["close"], color=BLUE, linewidth=1)
ax.set(ylabel="Цена при затваряне ($)", title="Apple, 2003-2007, коригирани цени")
plt.show()
````

След корекцията графиката показва истинския ход на цената. Почистването
се прави преди всичко друго: всеки пример за обучение, в който попада 28
февруари 2005 г., би показал на модела спад с 50%, който не се е случил.

Промяната в цената на акциите обаче почти не може да се предскаже само от
миналите цени. Днешната цена вече отразява това, което купувачите и
продавачите знаят до момента, включително всички минали цени. Цената се
влияе, когато се появи нова и неочаквана информация. На 9 януари 2007 г.
Apple представя първия iPhone и цената скача с 8% за един ден, а на
следващия ден с още 5%. В цените от предишните дни нищо не подсказва този
скок. Датата на представянето е била известна, но не и какво ще бъде
показано и как ще го приемат инвеститорите. Същото важи за тримесечните
отчети, които може да са по-добри или по-лоши от очакваното, за новите
продукти на конкурентите, съдебните дела, смяната на ръководството,
промените в лихвите и икономическите кризи. Преди да се случат, такива
събития не личат в цените, затова никой модел не може да ги предвиди само
от миналите цени.

### 3. Потреблението на ток

Потреблението на ток зависи от неща, които се повтарят: в работните дни
заводите и офисите работят, а през зимата се отоплява. Файлът
`data/france_electricity_daily.csv` съдържа потреблението на ток във Франция
за всеки ден от 2015 до 2019 г. в гигаватчасове (GWh). Данните са на RTE,
френския оператор на преносната мрежа. Във Франция много домове се
отопляват с ток, затова зимата личи силно. Проверките са същите като при
акциите: честота, липсващи дни, тренд и сезонност.

````python
import matplotlib.dates as mdates

energy = pd.read_csv(DATA_DIR / "france_electricity_daily.csv", parse_dates=["date"], index_col="date")["consumption_gwh"]
print("Дни:", len(energy), "| от", energy.index.min().date(), "до", energy.index.max().date())
print("Честота:", pd.infer_freq(energy.index), "| дни без стойност:", int(energy.asfreq("D").isna().sum()))
print("Средно за година (GWh):", energy.groupby(energy.index.year).mean().round().astype(int).to_dict())
print("Средно за месец (GWh):", energy.groupby(energy.index.month).mean().round().astype(int).to_dict())
WEEKDAYS = ["пон", "вт", "ср", "чет", "пет", "съб", "нед"]
by_weekday = energy.groupby(energy.index.dayofweek).mean().round().astype(int)
by_weekday.index = WEEKDAYS
print("Средно за ден от седмицата (GWh):", by_weekday.to_dict())

weeks = energy["2018-06-04":"2018-07-08"]
weekend = weeks.index.dayofweek >= 5
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), layout="constrained")
axes[0].plot(energy.index, energy, color=BLUE, linewidth=0.8)
axes[0].set(ylabel="Потребление за деня (GWh)", title="Франция, 2015-2019")
axes[1].plot(weeks.index, weeks, color="gray", linewidth=1)
axes[1].scatter(weeks.index[~weekend], weeks[~weekend], color=BLUE, s=20, zorder=3, label="работен ден")
axes[1].scatter(weeks.index[weekend], weeks[weekend], color=ORANGE, s=20, zorder=3, label="събота и неделя")
axes[1].set(title="Пет седмици през юни 2018 г.", ylim=(850, 1200))
axes[1].xaxis.set_major_locator(mdates.WeekdayLocator(byweekday=mdates.MO)) # деление на всеки понеделник
axes[1].xaxis.set_major_formatter(mdates.DateFormatter("%d.%m"))
axes[1].legend(loc="lower left", ncols=2)
plt.show()
````

Честотата е един ден (`D`) и липсващи дни няма. Тренд почти няма:
средното за година е около 1300 GWh. Сезонността е двойна. През януари
потреблението е средно 1687 GWh на ден, а през август 1023 GWh. В събота и
особено в неделя то е с 10-15% по-ниско, отколкото в работните дни.
Необичайни са празниците: тогава потреблението е почти като в неделя. И
двата ритъма се повтарят, затова миналото казва нещо за утре.

### 4. От ред към таблица: прозорци

Моделите от предишните части учат от таблица: всеки ред е един пример,
колоните са признаците, а етикетът е отделна колона. Времевият ред е една
дълга колона от стойности. За да стане таблица, той се нарязва на
**прозорци**: 20 поредни дни са признаците, а 21-вият ден е етикетът.
Следващият прозорец започва един ден по-късно, затова два съседни прозореца
имат 19 общи дни. Двадесет дни са почти три седмици, затова във всеки
прозорец всеки ден от седмицата се среща поне два пъти.

`sliding_window_view` от NumPy връща всички поредни парчета с дадена
дължина. Тук дължината е 21: 20 дни за признаци и един за етикет.

````python
WINDOW = 20 # колко поредни дни вижда моделът
windows = np.lib.stride_tricks.sliding_window_view(energy.to_numpy(), WINDOW + 1)

X = windows[:, :WINDOW] # ден 1 до ден 20 са признаците
y = windows[:, WINDOW] # ден 21 е етикетът
label_dates = energy.index[WINDOW:] # датата на етикета на всеки прозорец
print("Прозорци:", X.shape[0], "| признаци в прозорец:", X.shape[1])

columns = [f"ден {day}" for day in range(1, WINDOW + 1)] + ["етикет"]
display(pd.DataFrame(windows[:3].round().astype(int), columns=columns, index=label_dates[:3].rename("дата на етикета")))
````

Всеки ред на таблицата е един пример. Вторият прозорец е първият,
преместен с един ден: стойностите се местят с една колона наляво, а
етикетът на първия прозорец става ден 20 на втория. Анимацията показва
първите осем прозореца. Горе прозорецът се плъзга по реда, а долу
таблицата расте с по един ред. В рамка е потреблението на 21 януари:
етикет в първия ред, ден 20 във втория, ден 19 в третия.

### 5. Тренировъчни и тестови прозорци

Прозорците се разделят с `train_test_split`, както в предишните части, но
без разбъркване (`shuffle=False`): първите 80% са за обучение, а
последните 20% за проверка. При разбъркване до всеки тестов прозорец би
имало тренировъчни прозорци с почти същите дни, включително по-късни от
него. Моделът би се проверявал с дни, които отчасти вече е видял.
Истинската прогноза се прави само с миналото, затова и проверката е върху
прозорци, които са след всички тренировъчни. `train_test_split` разделя
наведнъж и датите на етикетите.

````python
X_train, X_test, y_train, y_test, dates_train, dates_test = train_test_split(
    X, y, label_dates, test_size=0.2, shuffle=False)
print("Обучение:", len(y_train), "прозореца, етикети от", dates_train[0].date(), "до", dates_train[-1].date())
print("Проверка:", len(y_test), "прозореца, етикети от", dates_test[0].date(), "до", dates_test[-1].date())
````

Тестовите етикети са почти цялата 2019 г. Първите тестови прозорци

### 6. Базови модели

Най-простата прогноза е **"утре = днес"**: етикетът се предсказва с
последния ден от прозореца. Втората е **"като преди седмица"**: етикетът се
предсказва с деня точно седмица по-рано, т.е. с ден 14. Това са **базовите
модели** при времевите редове: всеки модел трябва да е по-добър от тях,
иначе не е научил нищо полезно. `X_test[:, -1]` е последната колона (ден
20), а `X_test[:, -7]` е седмата колона отзад (ден 14).

Мярката за грешка е **средната абсолютна грешка** (MAE, mean absolute
error): средно с колко GWh прогнозата се разминава с истинското
потребление. `mean_absolute_error` от scikit-learn я изчислява.

````python
print(f'"Утре = днес": средна грешка {mean_absolute_error(y_test, X_test[:, -1]):.0f} GWh')
print(f'"Като преди седмица": средна грешка {mean_absolute_error(y_test, X_test[:, -7]):.0f} GWh')

naive_error = pd.Series(np.abs(y_test - X_test[:, -1]), index=dates_test)
by_day = naive_error.groupby(naive_error.index.dayofweek).mean().round().astype(int)
by_day.index = WEEKDAYS
print('Грешка на "утре = днес" по дни от седмицата (GWh):', by_day.to_dict())
````

"Утре = днес" греши най-много в понеделник и в събота, когато седмицата
сменя ритъма: в понеделник очаква неделно потребление, а в събота
делнично. "Като преди седмица" пази ритъма, но не следи промените от
седмица на седмица, например застудяване.

### 7. Линейна регресия

### 9. Ще помогнат ли RNN и LSTM?

**Рекурентните невронни мрежи** (RNN, recurrent neural networks) са
създадени за последователности. Те получават същите прозорци, но четат
дните един по един и пазят вътрешно състояние, т.е. памет за видяното
досега. Рекурентните мрежи са в последната тема на курса, "Архитектури на
невронни мрежи". Схемата е като фигура 10.12 в книгата An Introduction to
Statistical Learning, но с дните на прозореца.

<img src="images/rnn.png" width="960" alt="Схема на рекурентна мрежа. Вляво състоянието A има стрелка U обратно към себе си. Вдясно същата мрежа е разгъната по дните от 1 до 20: потреблението за всеки ден X влиза в състоянието с теглата W, състоянието минава към следващия ден с теглата U, а от последното състояние A20 с теглата B излиза прогнозата за ден 21.">

Вляво е кратката схема: стрелката U връща състоянието обратно в мрежата.
Вдясно същата мрежа е разгъната по 20-те дни. В ден 2 мрежата получава
потреблението X₂ и състоянието A₁ от ден 1 и изчислява новото състояние
A₂. За разлика от линейната регресия, която има отделно тегло за всеки от
20-те дни, мрежата използва едни и същи тегла W, U и B във всички дни.
Прогнозата за ден 21 се изчислява само от последното състояние A₂₀, което
е видяло всички 20 дни.

Информацията от ден 1 обаче минава през 19 преизчислявания на състоянието
и отслабва. При по-дълги прозорци простата RNN трудно се научава да
използва ранните дни. **LSTM** (long short-term memory) е вид рекурентна
мрежа, която освен състоянието пази и дългосрочна памет.

<img src="images/lstm.png" width="960" alt="Една стъпка на проста RNN и на LSTM. В RNN предишното състояние и потреблението за деня минават през един слой tanh. В LSTM горе минава дългосрочната памет C. Forget gate я умножава, през input gate към нея се добавят нови стойности, а output gate решава каква част от паметта става новото състояние A.">

Вляво е една стъпка на простата RNN: един слой tanh, както във формулата на
първата схема. Функцията tanh прилича на сигмоидната функция от
логистичната регресия, но дава числа между -1 и 1. Вдясно е една стъпка на
LSTM. Освен състоянието A, по горната линия от ден на ден минава
дългосрочната памет C. Трите gates σ са като логистичната регресия:
със сигмоидната функция изчисляват от X
и предишното състояние числа между 0 и 1. Тези числа се умножават (×) с
това, което минава през съответния gate: 0 спира всичко, а 1 пропуска всичко.

- **Forget gate** решава каква част от старата памет да остане.
- **Input gate** решава каква част от новите стойности да се добави (+)
  към паметта.
- **Output gate** решава каква част от паметта да стане новото
  състояние A.

По горната линия паметта се променя само с умножение и събиране, затова
информацията от първите дни може да стигне до последния ден.

За RNN и LSTM прозорците са същите, но всеки прозорец е последователност от 20
стъпки с по една стойност (потреблението за деня), а не ред от 20 отделни признака.
Затова масивът е тримерен, с форма (брой прозорци, 20 дни, 1 стойност на ден), 
тук (1806, 20, 1), а етикетът си остава потреблението на 21-вия ден.

Линейната регресия получи 20 дни и сама откри кои от тях са важни.
Рекурентната мрежа може да използва много по-дълги прозорци и да улови
по-сложни зависимости, например различен седмичен ритъм през зимата и през
лятото. Затова RNN и LSTM се използват за потребление на ток, текст, реч и
сигнали от сензори.

По-сложният модел обаче също не може да намери информация, която я няма в
прозореца. Никоя мрежа не може да познае от миналите 20 дни, че утре е 1
май. Празниците трябва да се подадат като отделен признак.

например OpenStreetMap. Тук ролята на такава услуга играе малка таблица с

към всяка обява, а проверката за `NaN` показва дали някой квартал липсва в
таблицата.

Разстоянието се смята в километри. Един градус географска ширина е около
111 km. Един градус дължина е 111 km, умножени по косинуса на ширината,

````python
neighborhoods = pd.read_csv(DATA_DIR / "sofia_neighborhoods.csv")
located = apartments.merge(neighborhoods, on="neighborhood", how="left")
print("Обяви без координати:", int(located["lat"].isna().sum()))

CENTER_LAT, CENTER_LON = 42.6977, 23.3219
north_km = (located["lat"] - CENTER_LAT) * 111
east_km = (located["lon"] - CENTER_LON) * 111 * np.cos(np.radians(CENTER_LAT))
located["distance_km"] = np.sqrt(north_km ** 2 + east_km ** 2)
located["price_per_m2"] = located["price_eur"] / located["area_m2"]
display(located[["address", "distance_km", "price_per_m2"]].head().round(1))

fig, ax = plt.subplots(figsize=(7, 4), layout="constrained")
ax.scatter(located["distance_km"], located["price_per_m2"], s=12, color=BLUE, alpha=0.6)
ax.set(xlabel="Разстояние до центъра (km)", ylabel="Цена на m² (€)", title="Цена на квадратен метър")
plt.show()
````

````python
homes_train, homes_test = train_test_split(located, test_size=0.2, random_state=42)
mean_prediction = np.full(len(homes_test), homes_train["price_eur"].mean())
print(f"Средна цена: средна грешка {mean_absolute_error(homes_test['price_eur'], mean_prediction):.0f} €")
for columns in (["area_m2"], ["area_m2", "distance_km"]):
    price_model = LinearRegression().fit(homes_train[columns], homes_train["price_eur"])
    error = mean_absolute_error(homes_test["price_eur"], price_model.predict(homes_test[columns]))
    print(f"{' + '.join(columns)}: средна грешка {error:.0f} €, коефициенти {price_model.coef_.round(0)}")
````

### 14. Признак, изчислен от цената

Грешката пада над два пъти, но резултатът е фалшив. `price_per_m2` е

### 20. Как да се подаде часът на модела

Има три начина:

### Какво е текстът?

Текстът е **последователност от символи**: букви, цифри, интервали и

Текстът е входът на много задачи:

- **класификация**: спам или не, положителен или отрицателен отзив, тема
  на статия;
- **търсене**: кои документи отговарят на заявка;
- **превод, обобщаване и отговори на въпроси**;
- **езикови модели и чатботове**, които предсказват следващата част от

Текстовете са с различна дължина, а моделът очаква вектор с фиксиран брой

### 21. Символи, думи, изречения

### 23. Разделяне на думи без библиотека

`split()` оставя препинателните знаци и главните букви: "FREE", "Free" и
"free!" са три различни думи. Функцията `tokenize` прави две неща: превръща
буквите в малки и заменя всеки символ, който не е буква или цифра, с
интервал. `isalnum()` проверява дали символът е буква или цифра и работи и
с кирилица.

````python
def tokenize(text):
    """Малки букви; всеки символ, който не е буква или цифра, става интервал."""
    cleaned = "".join(symbol.lower() if symbol.isalnum() else " " for symbol in text)
    return cleaned.split()


example = "FREE entry! Call now, txt WIN to 87121."
print("split:   ", example.split())
print("tokenize:", tokenize(example))
print("tokenize:", tokenize("Котката спи, а кучето гони!"))
print("Различни думи със split:", len({word for text in sms["text"] for word in text.split()}))
print("Различни думи с tokenize:", len({word for text in sms["text"] for word in tokenize(text)}))
````

### 24. Речник и дължина на вектора

Bag-of-Words представя текста чрез броя на думите в него. **Речникът** е
списък от думи, всяка със свой номер. Векторът на едно съобщение има по
едно число за всяка дума от речника: колко пъти думата се среща в
съобщението. Значи **векторът е дълъг колкото речника**.

Речникът се научава само от тренировъчните съобщения, както медианите и
мащабът в част 2. Затова първо съобщенията се разделят.

- `document_frequency` брои в колко съобщения се среща всяка дума;
- `build_vocabulary` подрежда думите по азбучен ред и дава на всяка номер;
- `vectorize` създава матрица с по един ред за съобщение и по една колона
  за дума и брои думите. Думите извън речника се пропускат.

````python
train_texts, test_texts, y_train, y_test = train_test_split(
    sms["text"], sms["label"], test_size=0.2, random_state=42, stratify=sms["label"]
)
train_tokens = [tokenize(text) for text in train_texts]
test_tokens = [tokenize(text) for text in test_texts]


def document_frequency(token_lists):
    """В колко текста се среща всяка дума."""
    frequency = {}
    for tokens in token_lists:
        for word in set(tokens):
            frequency[word] = frequency.get(word, 0) + 1
    return frequency


def build_vocabulary(token_lists, min_docs=1, stop_words=()):
    """Речник дума: номер на колона, без редките думи и без стоп думите."""
    frequency = document_frequency(token_lists)
    words = sorted(word for word, count in frequency.items()
                   if count >= min_docs and word not in stop_words)
    return {word: column for column, word in enumerate(words)}


def vectorize(token_lists, vocabulary):
    """Bag-of-Words: колко пъти всяка дума от речника се среща във всеки текст."""
    X = np.zeros((len(token_lists), len(vocabulary)), dtype=np.float32) # float32 заема половината памет на float64
    for row, tokens in enumerate(token_lists):
        for word in tokens:
            if word in vocabulary:
                X[row, vocabulary[word]] += 1
    return X


full_vocabulary = build_vocabulary(train_tokens)
print("Думи в речника, т.е. дължина на вектора:", len(full_vocabulary))
example_vector = vectorize([train_tokens[2]], full_vocabulary)[0]
print("Съобщение:", train_texts.iloc[2])
print("Ненулеви стойности:", np.count_nonzero(example_vector), "от", len(example_vector))
print({word: int(example_vector[column]) for word, column in full_vocabulary.items() if example_vector[column] > 0})
````

**разредено** представяне: при голям речник почти всички
стойности са нули. Всяко ново съобщение носи нови думи и речникът расте.
Затова част от думите се махат.

### 25. Кои думи да се махнат

Две групи думи могат да се махнат почти без загуба:

- **редките думи**, които се срещат само в едно тренировъчно съобщение.
  Това са имена, телефонни номера, правописни грешки. От един пример
  моделът не може да научи надеждно тегло;
- **стоп думите**: много чести думи, които служат за граматиката и се
  срещат еднакво често и в двата класа. Те не помагат да се различи спамът.

Готовите списъци със стоп думи съдържат думи като "i", "to" и "not". Дали
една дума е безполезна обаче се проверява с данните. Таблицата показва в
какъв дял от спам и от нормалните съобщения се среща всяка от най-честите
думи. Тук стоп дума е дума, която се среща в поне 5% от съобщенията и
чиито два дяла се различават най-много 1.5 пъти.

````python
frequency = document_frequency(train_tokens)
print("Думи само в едно съобщение:", sum(count == 1 for count in frequency.values()), "от", len(frequency))

spam_frequency = document_frequency([t for t, label in zip(train_tokens, y_train) if label == "spam"])
legit_frequency = document_frequency([t for t, label in zip(train_tokens, y_train) if label == "legit"])
n_spam, n_legit = int((y_train == "spam").sum()), int((y_train == "legit").sum())
frequent = sorted((word for word, count in frequency.items() if count >= 0.05 * len(train_tokens)),
                  key=frequency.get, reverse=True)
shares = pd.DataFrame({
    "% от спам": [100 * spam_frequency.get(word, 0) / n_spam for word in frequent],
    "% от legit": [100 * legit_frequency.get(word, 0) / n_legit for word in frequent],
}, index=frequent)
ratio = shares["% от спам"] / shares["% от legit"]
shares["стоп дума"] = (ratio >= 2 / 3) & (ratio <= 1.5)
display(shares.head(20).round(1))
````

````python
STOP_WORDS = set(shares.index[shares["стоп дума"]])
print("Стоп думи:", sorted(STOP_WORDS))
stop_count = sum(word in STOP_WORDS for tokens in train_tokens for word in tokens)
print(f"Дял на стоп думите от всички думи: {100 * stop_count / sum(len(t) for t in train_tokens):.0f}%")

vocabulary = build_vocabulary(train_tokens, min_docs=2, stop_words=STOP_WORDS)
print("Дължина на вектора: преди", len(full_vocabulary), "| след", len(vocabulary))
X_train = vectorize(train_tokens, vocabulary)
X_test = vectorize(test_tokens, vocabulary)
print("X_train:", X_train.shape, f"| памет {X_train.nbytes / 1e6:.0f} MB")
````

### 26. TF-IDF: теглото на думата зависи от рядкостта ѝ

Bag-of-Words брои думите и всички думи тежат еднакво. Махането на стоп
думите е грубо: думата или остава, или изчезва.
**TF-IDF** (term frequency, inverse document frequency) дава на всяка дума
тегло според това колко е рядка:

- **tf** е броят на думата в съобщението, както досега;
- **idf** = log(N / df), където N е броят на тренировъчните съобщения, а df
  е в колко от тях се среща думата.

Стойността в TF-IDF вектора е tf · idf. Дума, която е във всяко съобщение,
получава idf = log(1) = 0 и не влияе. Колкото по-рядка е думата, толкова
по-голямо е idf. df вече е изчислено в т. 25 с `document_frequency`.

````python
idf = {word: np.log(len(train_tokens) / count) for word, count in frequency.items()}
for word in ["i", "to", "you", "call", "free", "txt", "won"]:
    print(f"{word:5} в {frequency[word]:4} съобщения, idf {idf[word]:.2f}")

vocabulary_words = sorted(vocabulary, key=vocabulary.get) # думите по реда на колоните
idf_vector = np.array([idf[word] for word in vocabulary_words], dtype=np.float32)
tfidf_train = X_train * idf_vector # всяка колона се умножава по idf на своята дума
tfidf_test = X_test * idf_vector

row = 3670 # тренировъчно спам съобщение, в което две думи се срещат по два пъти
print("Съобщение:", train_texts.iloc[row])
example = pd.DataFrame({"брой": X_train[row], "idf": idf_vector, "TF-IDF": tfidf_train[row]}, index=vocabulary_words)
display(example[example["брой"] > 0].sort_values("TF-IDF", ascending=False).round(2).T)
````

"i" и "you" се срещат в над 1200 от 4128 съобщения и получават малко
idf, а "txt" и "won" са по-редки и тежат повече. В примера "prize" и
"cash" са почти еднакво редки (idf 4.32 и 4.34), но "prize" се среща два
пъти и е най-отпред с 2 · 4.32 = 8.64, дори пред по-рядката дума "worth".
"a" също е два пъти, но е честа дума и остава назад, без да е махната.
TF-IDF прави плавно това, което махането на стоп думите прави на ръка.

### 27. Косинусова близост: кои съобщения си приличат

Как да се намери тренировъчното съобщение, което най-много прилича на
дадено? Евклидовото разстояние зависи от дължината на текста: съобщение и
същото съобщение, написано два пъти, са далеч едно от друго, защото
броевете са двойни. **Косинусовата близост** сравнява
посоката на векторите, а не дължината им:

cos(x, z) = x·z / (‖x‖ ‖z‖)

Тя е 1 за вектори с еднаква посока и 0 за текстове без обща дума. Ако всеки
вектор се раздели на дължината си (**L2-нормализация**), косинусовата
близост е просто скаларното произведение. Стандартизацията в част 2
работеше по колони, а L2-нормализацията работи по отделните вектори.

````python
def l2_normalize(X):
    """Всеки ред се дели на дължината си. Редовете от нули остават нули."""
    lengths = np.linalg.norm(X, axis=1, keepdims=True)
    return X / np.where(lengths > 0, lengths, 1)


twice = vectorize([tokenize("free entry call now"), tokenize("free entry call now " * 2)], full_vocabulary)
print("Евклидово разстояние до същия текст, написан два пъти:", round(float(np.linalg.norm(twice[0] - twice[1])), 2))
print("Косинусова близост:", round(float(l2_normalize(twice)[0] @ l2_normalize(twice)[1]), 2))

unit_train, unit_test = l2_normalize(tfidf_train), l2_normalize(tfidf_test)
query = 238 # тестово спам съобщение
similarity = unit_train @ unit_test[query] # косинусовата близост до всяко тренировъчно съобщение
print("\nТестово съобщение:", test_texts.iloc[query])
for row in np.argsort(similarity)[::-1][:3]:
    print(f"{similarity[row]:.2f} {y_train.iloc[row]}: {train_texts.iloc[row]}")
for row in range(3):
    print(f"{similarity[row]:.2f} {y_train.iloc[row]}: {train_texts.iloc[row]}")
````

Трите най-близки тренировъчни съобщения са същата реклама, в която се
различава само телефонният номер. Косинусовата близост ги намира, макар
текстовете да не са еднакви. На нея се основава търсенето на сходни
документи. Тук тя показва и нещо за данните: спамът идва на вълни и почти
еднакви съобщения има и в обучението, и в проверката. Такива тестови
съобщения моделът разпознава лесно. Колко са, е задача за вкъщи.

### 29. Биграми

Прост начин да се запази част от реда е да се броят и **двойките съседни
думи**, наречени **биграми**. "кучето гони котката" дава думите "кучето",
"гони", "котката" и биграмите "кучето гони" и "гони котката". Отделните
думи се наричат **униграми**. Функцията `with_bigrams` добавя биграмите към
списъка с думи, а речникът и векторите се строят както досега.

````python
def with_bigrams(tokens):
    """Думите плюс всички двойки съседни думи."""
    return tokens + [f"{first} {second}" for first, second in zip(tokens, tokens[1:])]


print(with_bigrams(tokenize("кучето гони котката")))
print(with_bigrams(tokenize("котката гони кучето")))

bigram_train = [with_bigrams(tokens) for tokens in train_tokens]
bigram_vocabulary = build_vocabulary(bigram_train, min_docs=2, stop_words=STOP_WORDS)
X_train_bigrams = vectorize(bigram_train, bigram_vocabulary)
bigram_model = Perceptron(random_state=42).fit(X_train_bigrams, y_train)


def predict_bigrams(token_lists):
    """Предсказания на модела с биграми."""
    X = vectorize([with_bigrams(tokens) for tokens in token_lists], bigram_vocabulary)
    return bigram_model.predict(X)


bigram_predictions = predict_bigrams(test_tokens)
print("Дължина на вектора:", len(bigram_vocabulary), f"| памет {X_train_bigrams.nbytes / 1e6:.0f} MB")
print("Правилни: винаги legit", int((y_test == "legit").sum()), "| униграми", int((predictions == y_test).sum()),
      "| с биграми", int((bigram_predictions == y_test).sum()), "от", len(y_test))
print("Променени предсказания след разбъркване:", int((predict_bigrams(shuffled_tokens) != bigram_predictions).sum()))
````

Сега двете изречения имат различни биграми: "кучето гони" и "котката
гони". Моделът с биграми вижда реда на съседните думи и разбъркването
променя част от предсказанията. Цената е по-дълъг вектор и три пъти повече
памет. При спама точността почти не се променя, защото думи като "claim" и

случаите само защото спамът е рядък. Биграмите помагат там, където

### 31. Текстът като последователност: RNN и LSTM

Рекурентните мрежи от т. 9 работят и с текст. Текстът също е
последователност, само че вместо дни има думи (токени). Мрежата е същата,
а се сменят само две неща: какво влиза на всяка стъпка и какво излиза
накрая.

| | Ток (т. 9) | Текст |
|---|---|---|
| една стъпка | един ден | една дума (токен) |
| X на стъпката | 1 число: потреблението | вектор от m числа: вграждането на думата |
| дължина | 20 дни | L думи |
| форма на масива | (1806, 20, 1) | (брой текстове, L, m) |
| изход от последното състояние | потреблението на ден 21 | клас, например спам или legit |

Текстовете имат различна дължина, затова се изравняват до L думи:
по-дългите се отрязват, а по-късите се допълват с нули. Текстът става
поредица от вектори на три стъпки:

1. `tokenize` разделя текста на думи (т. 23).
2. Речникът дава на всяка дума номер (т. 24).
3. **Слоят за вграждане** (embedding layer) е таблица с по един ред от m
   числа за всяка дума от речника (т. 30). Номерът на думата избира реда.
   Таблицата се научава заедно с останалите тегла или се взема готова,
   например word2vec или GloVe.

Мрежата чете векторите отляво надясно с едни и същи тегла W и U.
Състоянието A след всяка дума обобщава текста дотук. От последното
състояние с теглата B и сигмоидната функция се изчислява вероятност за
спам, както при логистичната регресия. Всички тегла, включително
таблицата с вгражданията, се научават заедно с градиентно спускане.

Така редът на думите се запазва. Изреченията "кучето гони котката" и
"котката гони кучето" дават различни състояния, а при Bag-of-Words
векторите им са еднакви (т. 28). След "not" състоянието е различно, затова
следващата дума го променя по друг начин и мрежата може да научи, че "not"
обръща смисъла ѝ. В дълъг отзив простата RNN забравя началото по същата
причина, по която забравя първите дни в т. 9. Дългосрочната памет C и
gates на **LSTM** могат да пренесат например едно "not" или темата през
много думи.

Ако на всяка стъпка мрежата предсказва следващата дума, тя става **езиков
модел**. Задачата е същата като прогнозата за тока: етикетът е следващата
стъпка, тук следващата дума.

Колко помага това на практика? В книгата An Introduction to Statistical
Learning (раздел 10.5.1) с отзивите за филми от IMDb простата RNN познава
76% от тестовите отзиви, LSTM 87%, а моделът с Bag-of-Words 88%. При
класификация по тон Bag-of-Words трудно се побеждава. RNN и LSTM печелят
там, където редът е същността: превод, разпознаване на реч, генериране на
текст.

Съвременните езикови модели (като ChatGPT и Claude) използват
архитектурата Transformer. Тя не чете дума по дума, а гледа всички токени
наведнъж чрез **механизма на внимание** (attention). RNN, вниманието и
Transformer са в последната тема на курса, "Архитектури на невронни мрежи".

- **Дата и час:** низът става дата с `pd.to_datetime`, а от нея се вземат
  час, ден, месец и година. Часът е цикличен. Като число той почти не
  помага, а като кръг или като категория помага много.
- **Текст:** може да се раздели на символи, думи или изречения. Токенът е
  компромисът, който използват езиковите модели, а думите са по-простото
  решение.

**Bag-of-Words:**

1. `tokenize` прави буквите малки и маха препинателните знаци.
2. Речникът се научава от тренировъчните текстове. Дължината на вектора е
   броят на думите в речника.
3. Редките думи и стоп думите се махат и векторът става над два пъти
   по-къс. Стоп думите се проверяват с данните.
4. TF-IDF намалява теглото на честите думи, без да ги маха. Косинусовата
   близост сравнява текстове по посоката на векторите им, независимо от
   дължината.
5. Bag-of-Words губи реда на думите. Биграмите пазят реда на съседните
   думи с цената на много по-дълъг вектор.
6. Bag-of-Words не знае, че "good" и "great" са близки. Вгражданията
   научават такава близост от данните.
7. RNN и LSTM четат текста дума по дума като последователност от
   вграждания, както дните при тока, и така пазят реда на всички думи.
