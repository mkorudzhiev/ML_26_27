## Част 1. Pandas: таблици и оглед

**pandas** е библиотека за работа с таблични данни.

Основните обекти, които ще използваме, са:
- `DataFrame` - таблица от редове и колони;
- `Series`  една поредица от стойности, например една колона;
- `index` - етикетите на редовете.

В тази част ще:
- заредим CSV файл;
- създадем собствени `Series` и `DataFrame`;
- разгледаме структурата на таблицата;
- избираме редове и колони;
- работим с индекса;
- проверяваме типове, категории и липсващи стойности;
- филтрираме данни и ще направим графика.

### 1. Зареждане на данните от CSV файл

`pd.read_csv(...)` прочита CSV файл и връща `DataFrame`.
`shape` показва броя редове и колони, а `head()` показва първите редове.

````python
df = pd.read_csv("data/iris_with_missing.csv")
print("Форма (редове, колони):", df.shape)
display(df.head())
````

### 2. `DataFrame` и `Series`

- Цялата таблица е `DataFrame`.
- Една отделна колона обикновено е `Series`.
- Списък от имена на колони запазва резултата като `DataFrame`.

`df["sepal_length_cm"]` има една ос, а `df[["sepal_length_cm"]]` има две оси: редове и колона.

````python
sepal_length_series = df["sepal_length_cm"]
sepal_length_table = df[["sepal_length_cm"]]
lengths_table = df[["sepal_length_cm", "petal_length_cm"]]

print("sepal_length as Series:", type(sepal_length_series).__name__, sepal_length_series.shape)
display(sepal_length_series.head(3))
print("sepal_length as DataFrame:", type(sepal_length_table).__name__, sepal_length_table.shape)
display(sepal_length_table.head(3))
print("lengths as DataFrame:", type(lengths_table).__name__, lengths_table.shape)
display(lengths_table.head(3))
````

### Създаваме нови `Series` и `DataFrame`

Досега взехме колона от готова таблица. Можем да създадем и нови
обекти от стойности, които задаваме сами:

- `pd.Series([...], name="...")` създава едномерна поредица. `name` е нейното име.
- `pd.DataFrame({...})` създава таблица от речник: ключовете са имената
  на колоните, а стойностите са списъци или `Series` с данните им.
  Всяка колона трябва да съдържа по една стойност за всеки ред.

Следващите три цветя са само кратък пример, отделен от CSV файла.
Не задаваме индекс, затова pandas използва етикетите `0`, `1` и `2`.

````python
petal_lengths = pd.Series([1.4, 4.7, 5.6], name="petal_length_cm")
example_flowers = pd.DataFrame({
    "species": ["setosa", "versicolor", "virginica"],
    "petal_length_cm": petal_lengths,
    "light_conditions": ["слънце", "полусянка", "сянка"],
})

display(petal_lengths)
display(example_flowers)
print("Форма на вектора:", petal_lengths.shape)
print("Форма на таблицата:", example_flowers.shape)
````

````python
# ---ДА ОПИТАМЕ---
# При създаване от речник ключовете стават етикети на индекса.
# Преди да изпълним клетката, какъв index и shape очакваме?
flower_labels = pd.Series({
    "A": "setosa", "B": "versicolor", "C": "virginica"
}, name="species")
print("Индекс:", flower_labels.index.tolist())
print("Форма:", flower_labels.shape)
display(flower_labels)
````

### `columns`, `index` и `ndim`

Всеки `DataFrame` съдържа не само стойностите в таблицата, но и информация за нейната структура:

- **`columns`** - имената на колоните.
- **`index`** - етикетите на редовете.
- **`ndim`** - броят на измеренията на обекта. `DataFrame` има две измерения - редове и колони, затова `ndim` е `2` (това е голямата разлика с numpy ndarray).
- **`shape`** - размерът по всяко измерение. Например `(150, 5)` означава 150 реда и 5 колони.

`Series` е едномерна поредица от стойности, затова има `index`, но няма `columns`. Нейният `ndim` е `1`, а `shape` има само един размер, например `(150,)`.

Така можем да мислим за структурата на двата основни pandas обекта:

| | `DataFrame` | `Series` |
|---|---|---|
| `index` | етикети на редовете | етикети на стойностите |
| `columns` | имена на колоните | няма |
| `ndim` | `2` | `1` |
| `shape` | `(редове, колони)` | `(стойности,)` |

````python
print("Series ndim:", length_series.ndim)
print("DataFrame ndim:", length_table.ndim)
print("Series index:", length_series.index[:5].tolist())
print("DataFrame columns:", length_table.columns.tolist())
````

### 3. Индекс на редовете

Всеки `DataFrame` има **индекс (`index`)** - етикети, с които pandas идентифицира редовете.

- Ако не зададем индекс, pandas използва `0, 1, 2, ...`.
- Индексът е **етикет**, а не непременно позиция.
- Индексът може да съдържа числа, текст, дати и други стойности.
- Индексът не е задължително уникален.

````python
print("Първи индекси:", df.index[:10].tolist())
print("Тип на индекса:", type(df.index).__name__)
````

#### Позиция и етикет

След избор или пренареждане позицията на един ред може да се промени, но неговият индекс остава същият.

- `iloc` избира по **позиция**;
- `loc` избира по **етикет**.

````python
small = df.iloc[[2, 5, 10]]

display(small)
print("Позиции:", small.index.tolist())
````

````python
print("Ред с позиция 2:")
display(small.iloc[2])

print("Ред с етикет 2:")
display(small.loc[2])
````

````python
# ---ДА ОПИТАМЕ---
#
# А какъв ще е резултатът от това?
display(small.iloc[2:2])
display(small.loc[2:2])
````

#### Срезове с `loc` и `iloc`

- При `iloc` важи обичайното правило на Python: краят на среза **не се включва**.
- При `loc` крайният етикет **се включва**, ако съществува в индекса.

````python
display(df.loc[0:2, ["sepal_length_cm", "light_conditions", "species"]])
display(df.iloc[0:3, [0, 4, 5]])
````

#### Пренареждане на редовете

При обръщане на таблицата позициите се променят, но индексите остават свързани със същите редове.

````python
df_reversed = df.iloc[::-1]
display(df_reversed.head(3))
print("Първи индекси след обръщане:", df_reversed.index[:5].tolist())
````

````python
display(df_reversed.loc[2:0, ["sepal_length_cm", "light_conditions", "species"]])
display(df_reversed.iloc[0:3, [0, 4, 5]])
````

````python
# ---ДА ОПИТАМЕ---
#
# какво ще върне това?
# HINT: получава се интервал от позиция 2 до позиция 0
display(df_reversed.loc[147:149])
````

#### Собствен индекс

Съществуваща колона може да стане индекс чрез `set_index(...)`.
Тук използваме `species`, за да видим и че индексът може да има дубликати.

````python
df_by_species = df.set_index("species")
display(df_by_species.head())
display(df_by_species.index)
````

````python
display(df_by_species.loc["setosa"].head())
print("Брой редове с индекс 'setosa':", len(df_by_species.loc["setosa"]))
````

За разлика от първичен ключ в база данни, **индексът в pandas не е задължително уникален**. Затова изборът с `.loc[...]` може да върне повече от един ред.

````python
print("Типове:")
print(df.dtypes)
print()
print("Цветя по вид:")
print(df["species"].value_counts())
print()
print("Цветя по условия на осветяване:")
print(df["light_conditions"].value_counts(dropna=False))
````

## Част 2. Индексиране в NumPy и Pandas

Pandas надгражда идеите от NumPy с имена на колони и индекси на редове.

| NumPy | Pandas |
|---|---|
| `ndarray` | `DataFrame`, `Series` |
| избор основно по позиция | избор по позиция и по етикет |
| `array[ред, колона]` | `iloc[ред, колона]` |
| - | `loc[етикет, име_на_колона]` |

### 8. Позиции, етикети и форми

При NumPy `array[:, 0]` е едномерен масив, а `array[:, [0]]` е двумерен масив с една колона. Същата идея се вижда при `Series` и `DataFrame` в pandas.

````python
array = np.array([[10, 100], [20, 200], [30, 300], [40, 400]])
table = pd.DataFrame(array, columns=["x", "y"], index=["a", "b", "c", "d"]) # списъкът с етикети трябва да е с еднакъв размер като колоните

print("NumPy, позиции 1 и 2:", array[1:3, 0])
print("Pandas, позиции 1 и 2:")
display(table.iloc[1:3, [0]])
print("Pandas, етикети от b до d:")
display(table.loc["b":"d", ["x"]])
print("Форми:", array[:, 0].shape, array[:, [0]].shape)
````

#### От `DataFrame` към NumPy масив

`to_numpy()` връща стойностите като `ndarray`, без pandas индексите и имената на колоните.

````python
iris_array = df[numeric_features].to_numpy()
print(type(df[numeric_features]).__name__, df[numeric_features].shape)
print(type(iris_array).__name__, iris_array.shape)
````

### 9. Срезове: граници, стъпка и броене от края

Срезът `start:stop:step` избира последователни елементи. При NumPy и
`.iloc` работим с **позиции**: започваме от `start`, а позицията `stop`
**не влиза** в резултата. Ако пропуснем граница, продължаваме до
съответния край. `step` задава през колко позиции преминаваме;
отрицателната стъпка върви в обратен ред. Отрицателните позиции
броят отзад: `-1` е последната,
`-2` е предпоследната.

| Срез | Какво избира |
|---|---|
| `[1:-1]` | от втория елемент до предпоследния включително |
| `[-2:]` | последните два елемента |
| `[::2]` | всеки втори елемент, започвайки от първия |
| `[::-1]` | всички елементи в обратен ред |

При двумерен масив или таблица задаваме отделен избор за редовете и
колоните: `array[редове, колони]` и `table.iloc[редове, колони]`.

````python
print("NumPy, без първия и последния ред:", array[1:-1, 0])
print("NumPy, всеки втори ред:", array[::2, 0])
print("NumPy, редовете в обратен ред:", array[::-1, 0])
print("Pandas, последните два реда:")
display(table.iloc[-2:, :])
print("Pandas, всеки втори ред отзад напред:")
display(table.iloc[::-2, :])
````

#### Отрицателна позиция или отрицателен етикет?

При `.loc` числата са **етикети**, а не позиции. Крайната граница на
срез с `.loc` е включена. Например `.loc[-1:0]` включва редовете с
етикети `-1` и `0`, ако те са в индекса.

`load_iris(as_frame=True)` връща четирите оригинални измервания като Pandas таблица и вида като числов код. `target_names` съдържа имената на трите вида.

````python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
display(type(iris))
X_clean = iris.data
y_clean = iris.target

print("Признаци:", X_clean.shape, "цел:", y_clean.shape)
print("Липсващи измервания:", int(X_clean.isna().sum().sum()))
print("Имена на видовете:", iris.target_names.tolist())
display(X_clean.head(3))
````

### 11. Разделяне на данните за обучение и проверка

`X_clean` е двумерна таблица с признаци, а `y_clean` е едномерна поредица с целта.

`train_test_split` разделя редовете заедно от двата списъка:
- `test_size=0.2` оставя една пета за проверка;
- `random_state=42` прави разделянето възпроизводимо;
- `stratify=y_clean` запазва дела на трите вида и в двете части.

Моделът ще види само тренировъчните редове при `fit`.

````python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.2, random_state=42, stratify=y_clean
)
print("Тренировъчни:", X_train.shape, "тестови:", X_test.shape)
print("Видове в целия набор:", y_clean.value_counts().sort_index().to_dict())
print("Видове в тренировъчната част:", y_train.value_counts().sort_index().to_dict())
print("Видове в тестовата част:", y_test.value_counts().sort_index().to_dict())
````

`train_test_split` запазва оригиналните pandas индекси.

````python
print("Първи индекси в X_train:", X_train.index[:5].tolist())
print("Първи индекси в X_test:", X_test.index[:5].tolist())
````

### 12. Обучение и проверка

#### Два модела за сравнение

Ще сравним един много прост базов модел с модел, който действително използва характеристиките на цветята.

- **`DummyClassifier`** е умишлено прост модел, който използваме като **базова линия (baseline)**. Той не се опитва да открие връзка между характеристиките на ириса и неговия вид. При стратегия `most_frequent` по време на `fit()` моделът намира кой вид се среща най-често в тренировъчните данни, а при `predict()` предсказва този вид за **всеки** нов ирис, независимо от неговите размери. Ако един истински модел не се справя по-добре от тази проста стратегия, вероятно не сме научили нищо полезно от данните.

- **`KNeighborsClassifier` (KNN)** е модел, който използва характеристиките на тренировъчните примери. При предсказване за нов ирис той сравнява неговите размери с размерите на ирисите, които вече познава, намира **най-близките тренировъчни примери** и определя вида според техните видове. Параметърът `n_neighbors` задава колко близки примера да бъдат взети предвид. Например при `n_neighbors=5` моделът разглежда петте най-близки тренировъчни ириса и избира най-често срещания сред тях вид.

На този етап използваме KNN като готов алгоритъм от `scikit-learn`. По-важното е да проследим общия процес:
1. **създаваме модел**
2. **обучаваме го с `fit()`**
3. **правим предсказвания с `predict()`**
4. **оценяваме резултата**
5. **подобряваме**

Общият интерфейс е:
- `fit(X_train, y_train)` - обучение;
- `predict(X_test)` - прогнози;
- `accuracy_score(...)` - дял на правилните прогнози.

````python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier

baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
neighbor_model = KNeighborsClassifier(n_neighbors=3).fit(X_train, y_train)

baseline_predictions = baseline.predict(X_test)
predictions = neighbor_model.predict(X_test)
print("Точност на постоянната прогноза:",
      round(accuracy_score(y_test, baseline_predictions), 3))
print("Първите пет прогнози:", iris.target_names[baseline_predictions[:5]].tolist())
print("Първите пет действителни вида:", iris.target_names[y_test.iloc[:5].to_numpy()].tolist())
print("Точност на 3-NN:",
      round(accuracy_score(y_test, predictions), 3))
print("Първите пет прогнози:", iris.target_names[predictions[:5]].tolist())
print("Първите пет действителни вида:", iris.target_names[y_test.iloc[:5].to_numpy()].tolist())
````

#### Подобрение чрез промяна на хиперпараметри

Променяме `n_neighbors` и сравняваме точността на същото тестово множество.

````python
for k in [1, 3, 5, 10]:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    model_predictions = model.predict(X_test)
    print(f"k={k}: accuracy={accuracy_score(y_test, model_predictions):.3f}")
````

### Различни стратегии на `DummyClassifier`

`DummyClassifier` може да прави базови предсказвания по различни стратегии:

- **`most_frequent`** - винаги предсказва най-често срещания клас в тренировъчните данни.
- **`prior`** - използва честотите на класовете от тренировъчните данни. При `predict()` избира класа с най-голяма честота, затова предсказанията му обикновено съвпадат с `most_frequent`.
- **`stratified`** - избира класовете случайно, но вероятността за всеки клас следва честотата му в тренировъчните данни.
- **`uniform`** - избира всеки клас с еднаква вероятност. При три класа всеки има вероятност `1/3`.

Тези модели **не използват характеристиките на ирисите**, за да откриват зависимост между размерите и вида. Те ни дават различни прости базови резултати, с които можем да сравним истинския модел.

При случайните стратегии (`stratified` и `uniform`) конкретният резултат може да се различава между отделни изпълнения. Задаваме `random_state=42`, за да получаваме едни и същи случайни предсказвания при всяко изпълнение.

````python
strategies = ["most_frequent", "prior", "stratified", "uniform"]

for strategy in strategies:
    model = DummyClassifier(strategy=strategy, random_state=42)
    model.fit(X_train, y_train)

    predictions = model.predict(X_test)
    accuracy = accuracy_score(y_test, predictions)

    print(f"{strategy:13s}: accuracy = {accuracy:.2f}")
````
